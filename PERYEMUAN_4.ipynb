STACK DENGAN SINGLE LINKED LIST


```python
class Queue:
    def __init__(self):
        # Inisialisasi antrean menggunakan list [cite: 205, 206, 207]
        self.queue = list()

    def addtoqu(self, dataval):
        # Metode insert untuk menambahkan elemen [cite: 208]
        if dataval not in self.queue:
            # Menambahkan elemen di indeks 0 agar elemen lama bergeser ke kanan
            self.queue.insert(0, dataval)
            return True
        return False

    def removefromq(self):
        # Metode pop untuk menghapus elemen [cite: 211, 212]
        if len(self.queue) > 0:
            # Menghapus elemen paling belakang (elemen yang pertama kali masuk) [cite: 213, 214]
            return self.queue.pop()
        return ("No elements in Queue!")

    def size(self):
        # Mengembalikan jumlah item dalam antrean [cite: 216, 217]
        return len(self.queue)

# Inisialisasi objek antrean [cite: 218, 219]
TheQueue = Queue()

# Menambahkan elemen ke dalam antrean [cite: 220, 221, 222, 223]
print(TheQueue.addtoqu("Jan"))
print(TheQueue.addtoqu("Feb"))
print(TheQueue.addtoqu("March"))
print(TheQueue.addtoqu("April"))

# Menampilkan ukuran antrean [cite: 224]
print(TheQueue.size())

# Menghapus elemen dari antrean (mengikuti prinsip FIFO) [cite: 225, 226]
print(TheQueue.removefromq())
print(TheQueue.removefromq())

# Menampilkan ukuran antrean setelah penghapusan [cite: 227]
print(TheQueue.size())
```
