1. Representasi Graf dengan Adjacency List

```python
# Membuat graf tak berarah menggunakan adjacency list
graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'F'],
    'D': ['B'],
    'E': ['B', 'F'],
    'F': ['C', 'E']
}


# Menampilkan graf
def print_graph(graf):
    for node in graf:
        print(f"{node} -> {graf[node]}")


print_graph(graph)
```

2. Traversal Graf menggunakan BFS

```python
from collections import deque

def bfs(graph, start):
  visited = set()
  queue = deque([start])
  while queue:
    vertex = queue.popleft()
    if vertex not in visited:
      print(vertex, end=" ")
      visited.add(vertex)
      queue.extend([neighbor for neighbor in graph[vertex] if neighbor not in visited])
# Contoh pemanggilan
bfs(graph, 'A')
```

3. Traversal Graf menggunakan DFS

```python
def dfs(graph, start, visited=None):
  if visited is None:
    visited = set()
  if start not in visited:
    print(start, end=" ")
    visited.add(start)
    for neighbor in graph[start]:
      dfs(graph, neighbor, visited)
# Contoh pemanggilan
dfs(graph, 'A')
```

4. Latihan


```python
def count_neighbors(graph):
    result = {}
    for node in graph:
        # Menghitung panjang list dari daftar tetangga simpul tersebut
        result[node] = len(graph[node])
    return result

print("\n--- Jumlah Tetangga Tiap Simpul ---")
print(count_neighbors(graph))
```

```python
from collections import deque

# ==========================================
# SOAL 1: Buat graf tak berarah baru
# (Minimal 5 Simpul, 7 Sisi) [cite: 222]
# ==========================================
# Simpul: 'P', 'Q', 'R', 'S', 'T' (5 simpul)
# Sisi: (P-Q), (P-R), (Q-R), (Q-S), (R-S), (R-T), (S-T) [7 Sisi]

tugas_graph = {
    'P': ['Q', 'R'],
    'Q': ['P', 'R', 'S'],
    'R': ['P', 'Q', 'S', 'T'],
    'S': ['Q', 'R', 'T'],
    'T': ['R', 'S']
}

def print_graph(graf):
    for node in graf:
        print(f"{node} -> {graf[node]}")

print("=== JAWABAN TUGAS 1 (Graf Baru) ===")
print_graph(tugas_graph)
print("-" * 40)


# ==========================================
# SOAL 3: Modifikasi BFS agar Mengembalikan List
# Urutan Kunjungan [cite: 224]
# ==========================================
def bfs_list(graph, start):
    visited = set()
    queue = deque([start])
    visited_order = [] # List menampung urutan kunjungan

    while queue:
        vertex = queue.popleft()
        if vertex not in visited:
            visited_order.append(vertex)
            visited.add(vertex)
            queue.extend([neighbor for neighbor in graph[vertex] if neighbor not in visited])

    return visited_order


# ==========================================
# SOAL 2: Implementasi BFS & DFS dari Simpul Pilihan [cite: 223]
# (Menggunakan fungsi hasil modifikasi & fungsi DFS standar)
# ==========================================
def dfs_standar(graph, start, visited=None):
    if visited is None:
        visited = set()
    if start not in visited:
        print(start, end=" ")
        visited.add(start)
        for neighbor in graph[start]:
            dfs_standar(graph, neighbor, visited)

print("=== JAWABAN TUGAS 2 & 3 ===")
print("Jalur BFS (Mengembalikan List):", bfs_list(tugas_graph, 'P'))
print("Jalur DFS (Mencetak langsung): ", end="")
dfs_standar(tugas_graph, 'P')
print("\n" + "-" * 40)


# ==========================================
# SOAL 4: Fungsi find_path(graph, start, end)
# Menggunakan DFS [cite: 225]
# ==========================================
def find_path(graph, start, end, visited=None):
    if visited is None:
        visited = set()

    visited.add(start)

    # Jika simpul tujuan ditemukan, kembalikan jalurnya
    if start == end:
        return [start]

    for neighbor in graph[start]:
        if neighbor not in visited:
            path = find_path(graph, neighbor, end, visited)
            if path: # Jika jalur ke target valid (bukan None)
                return [start] + path

    return None

print("=== JAWABAN TUGAS 4 (Mencari Jalur) ===")
jalur_cari = find_path(tugas_graph, 'P', 'T')
print(f"Jalur dari 'P' menuju 'T': {jalur_cari}")
print("-" * 40)


# ==========================================
# SOAL 5: Fungsi is_connected(graph)
# Mengecek keterhubungan graf [cite: 226]
# ==========================================
def is_connected(graph):
    if not graph:
        return True

    # Ambil simpul pertama secara dinamis sebagai titik awal traversal
    start_node = list(graph.keys())[0]
    visited = set()

    # Inner function DFS untuk mendata semua simpul terjangkau
    def dfs_helper(node):
        visited.add(node)
        for neighbor in graph[node]:
            if neighbor not in visited:
                dfs_helper(neighbor)

    dfs_helper(start_node)

    # Jika total simpul yang dikunjungi sama dengan jumlah total simpul pada graf,
    # maka graf tersebut dinyatakan saling terhubung (Connected Graph).
    return len(visited) == len(graph)

print("=== JAWABAN TUGAS 5 (Cek Keterhubungan) ===")
print("Apakah tugas_graph saling terhubung?", is_connected(tugas_graph))

# Contoh pengujian pada Disconnected Graph
disconnected_graph = {
    'A': ['B'],
    'B': ['A'],
    'C': [] # Simpul terisolasi
}
print("Apakah disconnected_graph saling terhubung?", is_connected(disconnected_graph))
print("-" * 40)
```
