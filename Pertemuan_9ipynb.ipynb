1. Implementasi Mash Table Sederhana

```python
class SimpleHashTable:
    def __init__(self, size):
        self.size = size
        self.table = [None] * size

    def hash_function(self, key):
        return hash(key) % self.size

    def insert(self, key, value):
        index = self.hash_function(key)
        self.table[index] = value

    def search(self, key):
        index = self.hash_function(key)
        return self.table[index]

# Contoh penggunaan
ht = SimpleHashTable(10)
ht.insert("apel", 100)
ht.insert("pisang", 200)

print("Nilai untuk 'apel':", ht.search("apel"))
print("Nilai untuk 'pisang':", ht.search("pisang"))
```

Penanganan Collision dengan linear probing

```python
class LinearProbingHashTable:
    def __init__(self, size):
        self.size = size
        self.table = [None] * size

    def hash_function(self, key):
        return hash(key) % self.size

    def insert(self, key, value):
        index = self.hash_function(key)
        original_index = index

        while self.table[index] is not None:
            if self.table[index][0] == key:
                break
            index = (index + 1) % self.size
            if index == original_index:
                raise Exception("Hash table penuh")

        self.table[index] = (key, value)

    def search(self, key):
        index = self.hash_function(key)
        original_index = index

        while self.table[index] is not None:
            if self.table[index][0] == key:
                return self.table[index][1]
            index = (index + 1) % self.size
            if index == original_index:
                break
        return None

# Contoh penggunaan
ht = LinearProbingHashTable(5)
ht.insert("apel", 100)
ht.insert("pisang", 200)
ht.insert("melon", 300)

print("Nilai untuk 'apel':", ht.search("apel"))
print("Nilai untuk 'melon':", ht.search("melon"))
```

Penanganan Collision dengan chaining

```python
class ChainingHashTable:
    def __init__(self, size):
        self.size = size
        self.table = [[] for _ in range(size)]

    def hash_function(self, key):
        return hash(key) % self.size

    def insert(self, key, value):
        index = self.hash_function(key)
        for i, (k, v) in enumerate(self.table[index]):
            if k == key:
                self.table[index][i] = (key, value)
                return
        self.table[index].append((key, value))

    def search(self, key):
        index = self.hash_function(key)
        for k, v in self.table[index]:
            if k == key:
                return v
        return None

# Contoh penggunaan
ht = ChainingHashTable(5)
ht.insert("apel", 100)
ht.insert("pisang", 200)
ht.insert("melon", 300)

print("Nilai untuk 'apel':", ht.search("apel"))
print("Nilai untuk 'melon':", ht.search("melon"))
```

1. Eksperimen Sederhana Linear Probing

```python
class LinearProbingHashTable:
    def __init__(self, size):
        self.size = size
        self.table = [None] * size

    def hash_function(self, key):
        return hash(key) % self.size

    def insert(self, key, value):
        index = self.hash_function(key)
        original_index = index

        # Mencari slot kosong secara berurutan jika terjadi collision
        while self.table[index] is not None:
            if self.table[index][0] == key:
                break
            index = (index + 1) % self.size
            if index == original_index:
                raise Exception("Hash table penuh")

        self.table[index] = (key, value)

# a. Buat objek LinearProbingHashTable dengan ukuran tabel 5
lp_table = LinearProbingHashTable(5)

# b. Masukkan pasangan key-value yang ditentukan
data_lp = {"A": 10, "B": 20, "C": 30, "D": 40, "E": 50}
for key, val in data_lp.items():
    lp_table.insert(key, val)

# c. Cetak isi array self.table setelah semua data dimasukkan
print("Isi array self.table (Linear Probing):")
print(lp_table.table)
```

2. Percobaan Dasar Chaining

```python
class ChainingHashTable:
    def __init__(self, size):
        self.size = size
        # Membuat array yang berisi list kosong di setiap slotnya
        self.table = [[] for _ in range(size)]

    def hash_function(self, key):
        return hash(key) % self.size

    def insert(self, key, value):
        index = self.hash_function(key)
        # Cek apakah key sudah ada untuk diperbarui nilainya
        for i, (k, v) in enumerate(self.table[index]):
            if k == key:
                self.table[index][i] = (key, value)
                return
        # Jika belum ada, tambahkan (append) ke list di slot tersebut
        self.table[index].append((key, value))

# a. Gunakan ukuran tabel 3
chain_table = ChainingHashTable(3)

# b. Masukkan key-value yang ditentukan
data_chain = {"apel": 100, "melon": 200, "lemon": 300, "pisang": 400}
for key, val in data_chain.items():
    chain_table.insert(key, val)

# c. Cetak isi tabel dan tampilkan semua data yang disimpan di setiap slot
print("Isi tabel berdasarkan slot (Chaining):")
for i, slot in enumerate(chain_table.table):
    print(f"Slot {i}: {slot}")
```
