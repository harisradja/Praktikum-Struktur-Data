COIN CHANGE

```python
def coin_change_greedy(amount, coins):
  coins.sort (reverse = True)
  result= []
  for coin in coins:
    while amount >= coin:
      amount -= coin
      result.append(coin)
  return result

#contoh penggunaan
amount= 57
coins = [25, 10, 5, 1]
change = coin_change_greedy(amount, coins)
print("koin yang digunakan:", change)
```

MINIMUM SPANNING TREE: PRIM DAN KRUSKAL


```python
import heapq

def prim_mst(graph, start):
    visited = set()
    min_heap = [(0, start)]
    total_weight = 0
    while min_heap:
        weight, node = heapq.heappop(min_heap)
        if node in visited:
            continue
        visited.add(node)
        total_weight += weight
        for neighbor, edge_weight in graph[node]:
            if neighbor not in visited:
                heapq.heappush(min_heap, (edge_weight, neighbor))
    return total_weight

# Representasi graf: adjacency list (Sintaks dirapikan dari teks asli) [cite: 186, 187, 188, 192, 200, 209]
graph = {
    'A': [('B', 2), ('C', 3)],
    'B': [('A', 2), ('C', 1), ('D', 1)],
    'C': [('A', 3), ('B', 1), ('D', 4)],
    'D': [('B', 1), ('C', 4)]
}
print("Total bobot MST (Prim):", prim_mst(graph, 'A'))
```

```python
def kruskal_mst(edges, n_nodes):
    parent = {i: i for i in range(n_nodes)} # Diperbaiki menjadi dictionary comprehension [cite: 218]

    def find(x):
        while parent[x] != x:
            x = parent[x]
        return x

    def union(x, y):
        root_x = find(x)
        root_y = find(y)
        if root_x != root_y: # Diperbaiki dari teks asli "if root x 1 root y" [cite: 229, 230]
            parent[root_y] = root_x # [cite: 231]
            return True
        return False

    edges.sort(key=lambda x: x[2]) # [cite: 234]
    total_weight = 0
    for u, v, weight in edges:
        if union(u, v):
            total_weight += weight # [cite: 238]
    return total_weight

# edges: (node1, node2, weight), nodes are represented by integers
edges = [(0, 1, 2), (0, 2, 3), (1, 2, 1), (1, 3, 1), (2, 3, 4)] # [cite: 240]
print("Total bobot MST (Kruskal):", kruskal_mst(edges, 4))
```
