1. Membuat Binary Tree Secara Manual

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None

# Membuat tree secara manual
root = Node(1)
root.left = Node(2)
root.right = Node(3)
root.left.left = Node(4)
root.left.right = Node(5)

# Fungsi inorder traversal
def inorder(node):
    if node:
        inorder(node.left)
        print(node.data, end=' ')
        inorder(node.right)

print("Hasil inorder traversal dari binary tree:")
inorder(root)
```

2. MEMBUAT BINARY SEARCH


```python
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None

class BinarySearchTree:
    def __init__(self):
        self.root = None

    def insert(self, root, data):
        if root is None:
            return Node(data)
        if data < root.data:
            root.left = self.insert(root.left, data)
        else:
            root.right = self.insert(root.right, data)
        return root

# Inisialisasi dan pengisian data otomatis
bst = BinarySearchTree()
root = None
data_list = [50, 30, 70, 20, 40, 60, 80]

for value in data_list:
    root = bst.insert(root, value)
```

3. Traversal pada BST

```python
def inorder(node):
    if node:
        inorder(node.left)
        print(node.data, end=' ')
        inorder(node.right)

def preorder(node):
    if node:
        print(node.data, end=' ')
        preorder(node.left)
        preorder(node.right)

def postorder(node):
    if node:
        postorder(node.left)
        postorder(node.right)
        print(node.data, end=' ')

# Menampilkan hasil penelusuran
print("Inorder Traversal:")
inorder(root)
print("\nPreorder Traversal:")
preorder(root)
print("\nPostorder Traversal:")
postorder(root)
```

4. Pencarian Nilai dalam BST


```python
def search(node, key):
    # Basis: node tidak ditemukan atau data cocok dengan kunci
    if node is None or node.data == key:
        return node

    # Rekursif ke kiri jika kunci lebih kecil, ke kanan jika lebih besar
    if key < node.data:
        return search(node.left, key)
    return search(node.right, key)

# Uji pencarian angka 60
key = 60
result = search(root, key)
if result:
    print(f"{key} ditemukan dalam tree.")
else:
    print(f"{key} tidak ditemukan.")

# Uji pencarian angka 25
key = 25
result = search(root, key)
if result:
    print(f"{key} ditemukan dalam tree.")
else:
    print(f"{key} tidak ditemukan.")
```

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None

class BinarySearchTree:
    def __init__(self):
        self.root = None

    # Fungsi menyisipkan data sesuai aturan BST
    def insert(self, root, data):
        if root is None:
            return Node(data)
        if data < root.data:
            root.left = self.insert(root.left, data)
        else:
            root.right = self.insert(root.right, data)
        return root

    # Fungsi pencarian nilai
    def search(self, node, key):
        if node is None or node.data == key:
            return node
        if key < node.data:
            return self.search(node.left, key)
        return self.search(node.right, key)

# --- Implementasi Tugas ---
bst = BinarySearchTree()
root_bst = None

# Daftar 7 angka unik (NPM, TTL, ASCII Nama, Angka Bebas)
# Contoh data: [99, 15, 7, 65, 68, 10, 88]
data_list = [76, 21, 10, 114, 97, 10, 88]

for value in data_list:
    root_bst = bst.insert(root_bst, value)

# Menampilkan hasil traversal
def print_traversal(node, type="inorder"):
    if node:
        if type == "preorder": print(node.data, end=' ')
        print_traversal(node.left, type)
        if type == "inorder": print(node.data, end=' ')
        print_traversal(node.right, type)
        if type == "postorder": print(node.data, end=' ')

print("\nHASIL TRAVERSAL BST")
print("Inorder   :", end=' '); print_traversal(root_bst, "inorder")
print("\nPreorder  :", end=' '); print_traversal(root_bst, "preorder")
print("\nPostorder :", end=' '); print_traversal(root_bst, "postorder")

# Fitur Pencarian
print("\n\nHASIL PENCARIAN")
key_npm = 99
result = bst.search(root_bst, key_npm)
print(f"Mencari angka NPM ({key_npm}): {'Ditemukan' if result else 'Tidak Ditemukan'}")

key_false = 100
result_false = bst.search(root_bst, key_false)
print(f"Mencari angka bebas ({key_false}): {'Ditemukan' if result_false else 'Tidak Ditemukan'}")
```
